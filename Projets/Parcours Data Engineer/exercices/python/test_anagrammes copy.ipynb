```python
def annagrammes(mot1, mot2):
    
    mot1 = mot1.upper().strip()
    mot2 = mot2.upper().strip()
    
    mot1_complet = mot1
    mot2_complet = mot2
    
    if len(mot1_complet) == len(mot2_complet):
        print("Les 2 mots sont potentiellement des annagrammes")
    else:
        print("Les 2 mots ne sont pas des annagrammes car ils ont respectivement ", len(mot1_complet), " lettres contre ", len(mot2_complet))
        return 
    
    compteur = 1
    lettre_communes = []
    lettre_non_communes = []
    
    for lettre in mot1:
        if lettre in mot2:
            mot2 = mot2.replace(lettre, "", 1)
            lettre_communes.append(lettre)
            print(f"{compteur}. Lettre identique dans les 2 mots : ", lettre)
            compteur+=1
        else: 
            lettre_non_communes.append(lettre)
            print(f"{compteur}. Lettre non identique dans les 2 mots : ", lettre)

    if len(mot2) == 0:
        print(f"Les 2 mots {mot1_complet} et {mot2_complet} sont des anagrammes")
    else:
        # Tout ce qui reste dans mot2 à la fin sont les lettres en trop du deuxième mot
        lettre_non_communes_mot2 = list(mot2)
        
        print(f"Les 2 mots {mot1_complet} et {mot2_complet} ne sont pas des anagrammes. Voici le détail :")
        print(f"   - Lettres en trop dans {mot1_complet} : {lettre_non_communes}")
        print(f"   - Lettres en trop dans {mot2_complet} : {lettre_non_communes_mot2}")
        print(f"   - Lettres communes : {lettre_communes}")
    
```

```python
annagrammes("banane", "CaBaNe   ")
```
