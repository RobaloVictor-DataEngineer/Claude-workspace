```python
def annagrames(mot1, mot2):
    mot1 = mot1.replace(" ", "").upper()
    mot1_complet = mot1
    mot2 = mot2.replace(" ", "").upper()
    mot2_complet = mot2
    
    if len(mot1) != len(mot2):
        print("Mots pas de la même longueur, mot1 = ", len(mot1_complet), " mots et mot2 = ", len(mot2_complet), " mots, donc pas des annagrammes\n")
        return 0
    else:
        print(f"{mot1_complet} et {mot2_complet} sont de la même longueur, donc possiblement des annagrammes car ils contiennent tous les deux :", len(mot1), " lettres\n")
        
    diff_lettres1 = []
    
    for index, lettres in enumerate(mot1_complet):
        if lettres in mot2_complet:
            print(f"{index}. Lettre identiques {lettres} est dans {mot2}")
            mot2_complet = mot2_complet.replace(lettres, "", 1)
            
        else: 
            if lettres in mot2:
                print(f"{index}. Lettre en trop : '{lettres}' existe dans {mot2} mais n'est plus disponible (déjà utilisée)")
            else:
                print(f"{index}. Lettre absente : '{lettres}' n'existe pas du tout dans {mot2}")
        
            diff_lettres1.append(lettres)
        
    
    if len(mot2) == 0 :
        print(f"\nLes mots {mot1} et {mot2} sont des annagrames")
    else:
        print(f"\nLes mots {mot1} et {mot2} ne sont pas des annagrames")
        print("Liste des lettres non validées\n", diff_lettres1)
```
